```python
import pandas as pd
import numpy as np

df = pd.read_csv("ethiopia.csv")
df["Country"] = "Ethiopia"

# Fix date
df["date"] = pd.to_datetime(df["YEAR"]*1000 + df["DOY"], format="%Y%j")
df["Month"] = df["date"].dt.month

# Replace missing values
df.replace(-999, np.nan, inplace=True)

# Remove duplicates
print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()

# Missing values
print(df.isna().sum())
```

```python
monthly_temp = df.groupby("Month")["T2M"].mean()
monthly_temp.plot(title="Monthly Average Temperature")
```

```python
monthly_rain = df.groupby("Month")["PRECTOTCORR"].sum()
monthly_rain.plot(kind="bar", title="Monthly Rainfall")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```

Temperature is highest in March and April
Rainfall peaks during July–September
Temperature and humidity show a negative correlation
